```python
df[[
    "Quantity",
    "Unit_Price",
    "Discount",
    "Revenue",
    "Cost",
    "Profit",
    "Profit_Margin_%",
    "Shipping_Cost"
]].describe()
```

```python
print("Negative/Zero Quantity:", (df["Quantity"] <= 0).sum())
print("Negative Unit Price:", (df["Unit_Price"] <= 0).sum())
print("Invalid Discount:", ((df["Discount"] < 0) | (df["Discount"] > 1)).sum())
print("Negative Revenue:", (df["Revenue"] < 0).sum())
print("Negative Cost:", (df["Cost"] < 0).sum())
print("Negative Profit:", (df["Profit"] < 0).sum())
print("Negative Shipping Cost:", (df["Shipping_Cost"] < 0).sum())
```

```python
calculated_profit = df["Revenue"] - df["Cost"]

profit_difference = calculated_profit - df["Profit"]

print("Maximum difference:", profit_difference.abs().max())
print("Rows with difference > 0.01:", (profit_difference.abs() > 0.01).sum())
```

```python
print("Start Date:", df["Order_Date"].min())
print("End Date:", df["Order_Date"].max())
print("Number of Years:", df["Year"].nunique())
print("Years:", sorted(df["Year"].unique()))
```

```python
date_check = pd.DataFrame({
    "Year_from_Date": df["Order_Date"].dt.year,
    "Year_Column": df["Year"],
    "Month_from_Date": df["Order_Date"].dt.month,
    "Month_Column": df["Month"]
})

print("Year mismatches:", 
      (date_check["Year_from_Date"] != date_check["Year_Column"]).sum())

print("Month mismatches:", 
      (date_check["Month_from_Date"] != date_check["Month_Column"]).sum())
```
