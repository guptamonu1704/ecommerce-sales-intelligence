# 03. Business Insights & Recommendations

This section summarizes the key findings from the exploratory data analysis and translates them into actionable business insights and recommendations.

## 1. Overall Business Performance

```python
import pandas as pd

df = pd.read_csv("../Dataset/ecommerce_sales_dataset.csv")

df["Order_Date"] = pd.to_datetime(df["Order_Date"])

print("Data loaded successfully")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
print("Total Revenue:", round(df["Revenue"].sum(), 2))
print("Total Profit:", round(df["Profit"].sum(), 2))
print("Overall Profit Margin:", round((df["Profit"].sum() / df["Revenue"].sum()) * 100, 2), "%")
print("Total Orders:", df["Order_ID"].nunique())
```

## 2. Category Performance Insights

```python
category_analysis = (
    df.groupby("Category")
    .agg(
        Orders=("Order_ID", "count"),
        Revenue=("Revenue", "sum"),
        Profit=("Profit", "sum")
    )
)

category_analysis["Profit_Margin_%"] = (
    category_analysis["Profit"] / category_analysis["Revenue"] * 100
).round(2)

category_analysis = category_analysis.sort_values("Revenue", ascending=False)

category_analysis
```

```python
best_category_revenue = category_analysis["Revenue"].idxmax()
best_category_profit = category_analysis["Profit"].idxmax()
best_category_margin = category_analysis["Profit_Margin_%"].idxmax()

print("Highest Revenue Category:", best_category_revenue)
print("Highest Profit Category:", best_category_profit)
print("Highest Profit Margin Category:", best_category_margin)
```

## 3. Sub-Category Performance Insights

```python
subcategory_analysis = (
    df.groupby("Sub_Category")
    .agg(
        Orders=("Order_ID", "count"),
        Revenue=("Revenue", "sum"),
        Profit=("Profit", "sum")
    )
)

subcategory_analysis["Profit_Margin_%"] = (
    subcategory_analysis["Profit"] / subcategory_analysis["Revenue"] * 100
).round(2)

subcategory_analysis = subcategory_analysis.sort_values(
    "Revenue", ascending=False
)

subcategory_analysis
```

```python
print("Highest Revenue Sub-Category:",
      subcategory_analysis["Revenue"].idxmax())

print("Highest Profit Sub-Category:",
      subcategory_analysis["Profit"].idxmax())

print("Highest Profit Margin Sub-Category:",
      subcategory_analysis["Profit_Margin_%"].idxmax())
```

## 7. Key Business Findings

### Key Findings

- Electronics is the highest revenue and profit-generating category.
- Clothing has the highest profit margin among categories.
- Laptops are the strongest sub-category in terms of revenue and profit.
- Women's Wear has the highest profit margin among sub-categories.
- Middle East is the highest-revenue region and also has the highest regional profit.
- Regular customers contribute the highest revenue and profit.
- Female customers generate slightly higher revenue and profit than male customers.
- Delivered orders contribute the majority of orders, revenue, and profit.
- PayPal generates the highest revenue among payment methods, while Cash on Delivery has the highest profit margin.

## 8. Business Recommendations

### Recommendations

- Focus on Electronics and Laptops to sustain the strongest revenue and profit contribution.
- Explore opportunities to increase the contribution of high-margin categories such as Clothing.
- Strengthen marketing and retention strategies for Regular customers while developing New and Premium customers.
- Investigate the lower profit margin of VIP customers and identify opportunities to improve their profitability.
- Prioritize growth opportunities in the Middle East while maintaining performance across other regions.
- Monitor cancelled and returned orders to reduce potential revenue and profitability losses.
- Continue supporting multiple payment methods while evaluating high-margin methods such as Cash on Delivery and Google Pay.

## 9. Conclusion

The analysis shows a healthy overall business performance with total revenue of 5.28M, total profit of 1.44M, and an overall profit margin of 27.21%.

Electronics and Laptops are the strongest revenue and profit contributors, while Clothing and Women's Wear show strong profit margins. The Middle East is the leading region by revenue and profit, while Regular customers contribute the largest share of customer revenue and profit.

The business can improve further by focusing on high-performing products and regions, strengthening customer retention, improving VIP customer profitability, and reducing cancelled and returned orders.
